```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
  return x*x*x - x + 1

def fp(x):
  return 3*x*x - 1

def iter(x):
  return x - f(x)/fp(x)
```

```python
x = 1
x_ = np.linspace(-1.5, 1.5, 100) # Change -1.5 to your desired minimum x-value and 1.5 to your desired maximum x-value
```

```python
for _ in range(100):
  x = iter(x)
  plt.scatter(x,f(x))
  plt.plot(x_,f(x_))
  plt.plot(x_,0*x_,c='b')
  plt.xlim(-1.5, 1.5)
  plt.ylim(-4,4)
  #plt.show()
  print(x,f(x))
  if (abs(f(x)) < 1e-10):
    break
```

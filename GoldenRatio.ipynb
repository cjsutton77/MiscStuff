```python
# phi ^ 2 = 1 + phi

epsilon = 1e-15
```

```python
def golden(phi):
  return phi * phi - phi - 1
```

```python
high = 10
low = 1
phi = (high + low) / 2
iter = 0
while(True):
  iter += 1
  if abs(golden(phi)) < epsilon:
    break
  elif (golden(phi) < 0):
    low = phi
    phi = (high + low) / 2
  elif (golden(phi) > 0):
    high = phi
    phi = (high + low) / 2
print(f'Tolerance: {epsilon}')
print(f'iteration: {iter} phi: {phi} value: {golden(phi)}')
```
